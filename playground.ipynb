```python
import os
import cv2
import numpy as np
import matplotlib.pyplot as plt
stimuli_dir = '/ptmp/hleosme/avr_stimuli' 
output_dir = '/ptmp/hleosme/avr_output'
```

```python
idx = 230
chunk_size = 100
idx % chunk_size
```

```python
frames_direc = os.path.join(stimuli_dir, 'video_frames_30FPS')
img_sizes = [os.path.getsize(os.path.join(frames_direc, stimulus)) for stimulus in os.listdir(frames_direc)]
img_sizes = np.array(img_sizes) / 1000
# Bytes to kilobytes conversion
plt.hist(img_sizes, bins=50)
plt.xlabel('Image Size (KB)')
plt.ylabel('Frequency')
plt.title('Distribution of Image Sizes')
plt.show()

mean_size = np.mean(img_sizes)
median_size = np.median(img_sizes)
std_size = np.std(img_sizes)
print(f"Mean Size: {mean_size:.2f} KB, Median Size: {median_size:.2f} KB, Standard Deviation: {std_size:.2f} KB (N={img_sizes.size})")
```
